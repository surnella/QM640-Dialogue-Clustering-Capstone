# 01. Exploratory Data Analysis & Canonical Cleaning

This notebook performs data ingestion, canonical mapping, missing value analysis, and distributional analysis for Microsoft Frames, MultiWOZ 2.2, and Enterprise golden dialogues.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import json

# Load summary dataframe
df_summary = pd.read_csv("data/canonical/dialogue_summary.csv")
print("Summary shape:", df_summary.shape)
df_summary.head()
```

```python
df_summary.groupby("source_dataset")[["total_turns", "user_turns", "wizard_turns", "revisions_count", "hesitation_markers_count"]].describe().T
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
sns.histplot(df_summary["total_turns"], bins=25, kde=True, ax=axes[0], color="#2b5c8f")
axes[0].set_title("Distribution of Total Turns")
sns.boxplot(data=df_summary, x="source_dataset", y="total_turns", ax=axes[1], palette="Set2")
axes[1].set_title("Total Turns by Dataset Source")
plt.tight_layout()
plt.show()
```
